```python
import numpy as np
import pandas as pd

import scistudio

# Read the input port. `input` returns whatever this run bound to the
# port; `load` turns it into a SciStudio object. The same two lines work
# in the session and in the packaged block -- that is the whole point.
readings = scistudio.load(scistudio.input("readings"))
df = readings.to_pandas()
df.head()
```

```python
# Rebind rather than mutate in place. `df.dropna(...)` returns a new
# frame and the assignment is what the dependency analysis sees; an
# in-place `df.dropna(inplace=True)` changes the object with no edge to
# show for it, and a cell below would go stale without being marked.
df = df.dropna(subset=["intensity"])
```

```python
# Work the computation out one cell at a time. This is why the notebook
# shape exists: nobody knew the threshold before looking at the data.
threshold = float(df["intensity"].median()) * 1.5
threshold
```

```python
peaks = df[df["intensity"] > threshold].reset_index(drop=True)
len(peaks)
```

```python
# Declare the outputs. In the session this only registers the names --
# packaging is what turns a declaration into a port. In the packaged
# block the same call writes each object to its output folder.
#
# `output` takes native objects: the pandas frame below is wrapped into
# a DataFrame by construction at this boundary, not by you.
scistudio.output(peaks=peaks)
```
