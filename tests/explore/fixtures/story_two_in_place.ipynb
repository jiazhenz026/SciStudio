```python
import numpy as np
import pandas as pd

import scistudio

df = pd.DataFrame(
    {
        "a": [1.0, 2.0, None, 4.0, 5.0],
        "intensity": [10.0, 250.0, 300.0, 120.0, 400.0],
    }
)
values = np.array([3.0, 250.0, 300.0, 120.0, 400.0])
```

```python
df = df[df.intensity > 100].copy()
```

```python
df.dropna(inplace=True)
```

```python
peaks = [round(float(df["intensity"].sum()), 6), round(float(values.sum()), 6)]
```

```python
df.head()
```

```python
scistudio.output(peaks=peaks, table=df, signal=values)
```
